```python
import pandas as pd

cat_with_names = pd.read_csv("data/categories_with_names.csv")
cat_with_names.head()
```

```python
prod_with_cat_names = prod_orig.join(
    cat_with_names.set_index('category_id')[['category_name']],
    on='category_id',
    how='left'
)
prod_with_cat_names.head()
```

```python
prods = prod_with_cat_names.drop_duplicates('product_id').sort_values('product_id').reset_index(drop=True)[['product_id','category_code', 'category_name']]
prods.head()
```

```python
prods.shape
```

```python
prods.to_csv("data/products_with_category_names.csv", index=False)
```

```python
prods_with_names = pd.read_csv("data/products_with_product_names.csv")
prods_with_names.head()
```

```python
prod_with_prod_names = prod_with_cat_names.join(
    prods_with_names.set_index('product_id')[['product_name']],
    on='product_id',
    how='left'
)
prod_with_prod_names.head()
```

```python
prod_with_prod_names[['product_id', 'product_name', 'category_id', 'category_name', 'category_code', 'brand']].drop_duplicates('product_id').reset_index(drop=True).to_csv("data/products.csv", index=False)
```

```python
user_ids = pd.read_csv("data/user_ids.csv")
user_ids.head()
```

```python
current_len = len(user_ids)

# docelowa długość
target_len = 5_000_000

missing = target_len - current_len

# tworzymy nowe wartości
start_value = 595_414_563 + 1   # kolejna po ostatniej
new_values = pd.DataFrame({
    "user_id": range(start_value, start_value + missing)
})

# łączymy z obecnym dataframe
user_ids_extended = pd.concat([user_ids, new_values], ignore_index=True)
user_ids_extended.shape
```

```python
user_ids_extended.sort_values("user_id").reset_index(drop=True).to_csv("data/user_ids_extended.csv", index=False)
```

```python
user_ids_extended = pd.read_csv("data/user_ids_extended.csv")
user_ids_extended.head()
```

```python
users = pd.concat([user_ids_extended, users_main], axis=1)
users.head()
```

```python
users.to_csv("data/users.csv", index=False)
```
